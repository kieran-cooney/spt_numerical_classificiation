# Definitions

```python
L = 20
```

```python
def get_hamiltonian(L, t):
    terms = [
        ['zxz', [[-1*t, i, (i+1)%L, (i+2)%L] for i in range(L)]],
        ['x', [[-1*(1-t), i] for i in range(L)]],
    ]

    basis = spin_basis_1d(L)

    h = hamiltonian(
        terms,
        [],
        basis=basis,
        dtype=np.complex128,
        check_symm=False,
        check_herm=False
    )

    return h
```

```python
parameters = np.linspace(0, 1, 21)
```

## 20 site

```python
L = 20
```

```python
for t in tqdm(parameters):
    h = get_hamiltonian(L, t)
    e, psi = h.eigsh(k=1, which='SA')

    idx = np.argsort(e)
    e = e[idx]
    psi = psi[:, idx]

    t_string = str(int(100*t))
    np.savez(rf'../data/quspin_transverse_cluster_20_site/{t_string}.npz', energy=e, psi=psi)
```
